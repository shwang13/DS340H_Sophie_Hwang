## Interaction Plot 

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#read csv file
weekly = pd.read_csv('bluebike_weekly.csv', keep_default_na=True)
```

```python
lm = sns.lmplot(data=weekly, x="new_covid_cases", y="count_per_week", hue="season", scatter=True, palette='rocket', height=4, aspect=1.5)
```

```python
lm = sns.lmplot(data=weekly, x="tavg", y="count_per_week", hue="season", scatter=True, palette='rocket', height=4, aspect=1.5)
```

```python
weekly['week_start'] = pd.to_datetime(weekly['week_start'])
weekly = weekly.sort_values('week_start')
sns.lineplot(data=weekly, x='week_start', y='count_per_week')
```
